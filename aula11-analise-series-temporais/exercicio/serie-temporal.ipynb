Funções de autocorrelação vistas em aula (o Prophet não tem funções de ACF). Execute a célula abaixo para poder usá-las no exercício.

````python
def acf(x, nlags):
    """Autocorrelação r_k para k = 0..nlags."""
    x = np.asarray(x, dtype=float)
    x = x - x.mean()
    denominador = np.sum(x ** 2)
    return np.array([np.sum(x[k:] * x[:len(x) - k]) / denominador for k in range(nlags + 1)])


def plot_acf(x, lags, ax, titulo='ACF'):
    """Correlograma com intervalo de confiança de 95% (área azul, fórmula de Bartlett)."""
    n = len(x)
    valores = acf(x, lags)
    k = np.arange(lags + 1)
    limite = 1.96 * np.sqrt((1 + 2 * np.concatenate([[0, 0], np.cumsum(valores[1:-1] ** 2)])) / n)
    limite[0] = np.nan
    ax.fill_between(k, -limite, limite, alpha=0.25, label='IC 95%')
    ax.vlines(k, 0, valores)
    ax.plot(k, valores, 'o', markersize=4)
    ax.axhline(0, color='black', linewidth=0.8)
    ax.set_ylim(-1.1, 1.1)
    ax.set_xlabel('Lag (meses)')
    ax.set_title(titulo)
    return valores
````
